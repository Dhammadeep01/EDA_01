##### minmax Scaler.

mms =    (arr - arr.min)  /   (arr.max - arr.min)

```python
a = [ 1, 2, 3, 4, 5, 6, 7]    # rupeess 
b = [ 10, 20, 30, 40, 50, 60, 70]    #  paise    1 rupess = 10 paise
c = [ 100, 200, 300, 400, 500, 600, 700]  # 
```

```python
def  mms(arr):
    return   ( arr -  arr.min())  /  (arr.max() - arr.min())   # bodmas
```

```python
mms( np.array(a))
```

```python
mms( np.array(b))
```

```python
mms(np.array(c))
```
